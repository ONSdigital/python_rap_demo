```python
import pandas as pd

# Load data
df = pd.read_csv("../data/input/health_data.csv")

print("\nFirst 5 rows of cleaned data:")
print(df.head())
```

## Step 1: Review the Monolithic Script

Below is a single script that identifies missing values and performs imputation. Your task is to refactor this into modular code.

**Tasks:**
- Convert the below into a functions
- Extend the function to impute missing height and weight using group means (e.g., by gender or diagnosis)
- Add a new column to flag which values were imputed

**Hints:**
- Use clear function names and docstrings
- Use `groupby` and `transform` to calculate group means
- Use boolean indexing to flag imputed values

```python
# Impute missing height and weight with column means
df = df.copy()

# Flag missing values before imputation
df["height_cm_imputed"] = df["height_cm"].isna()
df["weight_kg_imputed"] = df["weight_kg"].isna()

# Impute using group means
df["height_cm"] = df["height_cm"].fillna(df.groupby("gender")["height_cm"].transform("mean"))
df["weight_kg"] = df["weight_kg"].fillna(df.groupby("gender")["weight_kg"].transform("mean"))

# Handle edge case: if all values in a group are missing, fallback to overall mean
df["height_cm"] = df["height_cm"].fillna(df["height_cm"].mean())
df["weight_kg"] = df["weight_kg"].fillna(df["weight_kg"].mean())

print("\nFirst 5 rows of cleaned data:")
print(df.head())
```
