# Exercise: Using modules and functions

In this exercise, you'll practice refactoring a single, monolithic script into a modular pipeline using functions and separate files. This is a key RAP best practice. Modules and functions ensure:
- Code is not duplicated unnecessarily
- The pipeline can be scaled up easily by using existing functions or adding new modules when needed
- Code is easy to navigate through the main.py file and appropriately named modules
- Code is easy to understand with comprehensive documentation for each function

**Learning objectives:**
- Understand the difference between monolithic and modular code
- Practice breaking up a large script into reusable functions
- Learn how to organise code into modules for analysis and reporting
- Learn how to handle missing data using imputation


```python
# Setup code for exercises

import pandas as pd

# Load data
df = pd.read_csv("../data/input/health_data.csv")

print("\nFirst 5 rows of cleaned data:")
print(df.head())
```

## Step 1: Review the Monolithic Script

Below is a single script that identifies missing values and performs imputation. Your task is to refactor this into modular code.

```python
# Impute missing height and weight with column means
df = df.copy()

# Flag missing values before imputation
df["height_cm_imputed"] = df["height_cm"].isna()
df["weight_kg_imputed"] = df["weight_kg"].isna()

# Impute using group means
df["height_cm"] = df["height_cm"].fillna(df.groupby("sex")["height_cm"].transform("mean"))
df["weight_kg"] = df["weight_kg"].fillna(df.groupby("sex")["weight_kg"].transform("mean"))

# Handle edge case: if all values in a group are missing, fallback to overall mean
df["height_cm"] = df["height_cm"].fillna(df["height_cm"].mean())
df["weight_kg"] = df["weight_kg"].fillna(df["weight_kg"].mean())

print("\nFirst 5 rows of cleaned data:")
print(df.head())
```
