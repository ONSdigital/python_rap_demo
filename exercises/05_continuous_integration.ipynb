# Continuous Integration Exercises

This set of exercises will help you understand and implement continuous integration (CI) best practices in your RAP pipeline. You will learn how to add, configure, and test CI elements using files and folders already present in this repository.

## Exercise 1: Explore the CI configuration files

Open and review the following files in the repository:
- `.pre-commit-config.yaml`: Pre-commit hooks for code quality, notebook formatting, secrets detection, and more.
- `pyproject.toml`: Configuration for tools like black, isort, flake8, pytest, and bandit.
- `requirements.txt`: Lists required Python packages for the pipeline and CI tools.

**Task:**  
Read through these files and note which tools and hooks are enabled.  
Which hooks check notebooks? Which check Python scripts?

## Exercise 2: Add a new pre-commit hook and test with a deliberate error

Add a new hook to `.pre-commit-config.yaml` to check for Python syntax errors before committing.  
You can use the `check-yaml` or `check-ast` hook from `pre-commit-hooks`.

**Task:**  
1. Add the following to your `.pre-commit-config.yaml` under the `pre-commit-hooks` repo:
   ```
   - id: check-ast
     name: Check Python files for syntax errors
     exclude: ^docs/user_guide/user_documentation\.md$
   ```
2. Save the file and run:
   ```
   pre-commit install
   pre-commit run --all-files
   ```
3. Add a deliberate syntax error to any Python file (e.g., `def broken_func(`).
4. Run the hook again:
   ```
   pre-commit run check-ast --all-files
   ```
5. Observe the output and note the error reported. Then fix the error and rerun the hook to confirm it passes.

These pre-commits should also run when you commit code using `git commit` or committing through your code editor.

## Exercise 3: Detect secrets in your code

The `detect-secrets` hook scans for sensitive information.

**Task:**  
1. Add a fake secret (e.g., `password = "12345"`) to any Python file.
2. Run:
   ```
   pre-commit run detect-secrets --all-files
   ```
3. Remove the secret and rerun the hook to confirm it passes.

## Exercise 4: Add a new workflow that activates on push to any branch

Create a new GitHub Actions workflow file in `.github/workflows/` (e.g., `ci_lint_and_format_any_branch.yml`).

**Task:**
1. Add the following workflow YAML to `.github/workflows/ci_lint_and_format_any_branch.yml`:
   ```yaml
   # This workflow runs linting and formatting checks on every branch push
   name: Lint and Format CI (Any Branch)
   on:
     push:
       branches: [ "*" ]  # Trigger on push to any branch
   jobs:
     lint-and-format:
       runs-on: ubuntu-latest  # Use the latest Ubuntu runner
       steps:
         - uses: actions/checkout@v4  # Check out the repository code
         - name: Set up Python
           uses: actions/setup-python@v5
           with:
             python-version: '3.11'  # Use Python 3.11
         - name: Install dependencies
           run: |
             python -m pip install --upgrade pip  # Upgrade pip
             pip install -r requirements.txt      # Install required packages
         - name: Run Black (code formatter)
           run: |
             black --check src/ tests/  # Check code formatting in src/ and tests/
         - name: Run Flake8 (linter)
           run: |
             flake8 src/ tests/         # Run linting in src/ and tests/
   ```
2. Commit and push your change to any branch.
3. Go to your repository on GitHub, click on the **Actions** tab, and look for the workflow run named "Lint and Format CI (Any Branch)".
4. Check the status and logs of the workflow to confirm it ran successfully.

This action will run any time you push code up to GitHub on any branch.

## Reflection

Continuous integration automates code quality checks, formatting, secrets detection, and testing.  
By following these exercises, you ensure your RAP pipeline is reproducible, robust, and easy to maintain.
