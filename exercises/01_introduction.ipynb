# Introduction to RAP pipeline exercises

Welcome to the RAP (Reproducible Analytical Pipeline) exercises for this project. These exercises are designed to help you learn and apply best practices for reproducible, modular, and transparent data analysis in Python.

## Contents of the exercises
- **01_introduction.ipynb**: Overview and guidance for the RAP exercises
- **02_modules.ipynb**: Refactor monolithic code into functions and modular scripts
- **03_config_files.ipynb**: Use configuration files to control pipeline behaviour
- **04_unit_tests.ipynb**: Write and run unit tests for your code
- **05_logging.ipynb**: Add logging for transparency and debugging
- **06_continuous_integration.ipynb**: Implement and test continuous integration

## Aim of the exercises
The aim is to guide you through building a robust, reproducible analytical pipeline using Python. You will:
- Learn to write modular, maintainable code
- Use configuration files for flexibility
- Implement automated testing and CI/CD
- Document and visualise your analysis

## Where to find exercises, solutions, and outputs
- **Exercises**: Located in the `exercises/` folder (e.g., `exercises/02_modules.ipynb`)
- **Solutions**: Located in `exercises/solutions/` (e.g., `exercises/solutions/02_modules_solutions.ipynb`)
- **Outputs**: Saved in `exercises/outputs/` (e.g., cleaned data, reports, charts)

## How to use the exercises
1. **Read each exercise notebook** and follow the instructions step by step.
2. **Write your code** in the notebook cells or in the `src/python_rap_demo/` modules as instructed.
4. **Check your solutions** by comparing your results with the solutions notebooks in `exercises/solutions/`.
5. **View outputs** in the `data/outputs/` folder (e.g., cleaned data, markdown reports, charts).

## Adding code to the src folder
- Place reusable functions and classes in the appropriate module in `src/python_rap_demo/` (e.g., `cleaning.py`, `io.py`, `report.py`).
- Use clear function names, type hints, and docstrings following PEP8 standards.

## How to check if your solutions have worked
- **Compare outputs**: Check your results against the solutions notebooks and output files.
- **Run tests**: Use the provided unit tests in the `tests/` folder. You can run all tests with:
  ```cmd
  pytest tests
  ```
- **Check outputs**: Review generated files in `data/outputs/` for expected results.
- **Use pre-commit hooks and CI**: Ensure your code passes formatting, linting, and CI checks.

By following these exercises, you will build a reproducible, automated, and transparent analytical pipeline suitable for real-world data analysis.
