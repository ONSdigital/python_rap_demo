# Introduction to RAP pipeline exercises
Welcome to the RAP (Reproducible Analytical Pipeline) exercises. These exercises are designed to help you develop understanding of best practices for a RAP and walk through how to apply those practices in Python.

## Contents of the exercises
- **01_introduction.ipynb**: Overview and guidance for the RAP exercises
- **02_modules.ipynb**: Refactor monolithic code into functions and modular scripts
- **03_config_files.ipynb**: Use configuration files to control pipeline behaviour
- **04_unit_tests.ipynb**: Write and run unit tests for your code
- **05_continuous_integration.ipynb**: Implement and test continuous integration

## Aim of the exercises
The aim is to guide to build on your understanding of RAP with practical experience in a demonstration repository.  
  
These exercises are intended as a starting point that you can build upon. Please use this repository to practice elements of RAP you would like to improve that may not be covered by these exercises.

## Where to find exercises, solutions, and outputs
- **Exercises**: Located in the `exercises/` folder (e.g., `exercises/02_modules.ipynb`)
- **Solutions**: Located in `exercises/solutions/` (e.g., `exercises/solutions/02_modules_solutions.ipynb`)
- **Outputs**: Saved in `data/outputs/` or `reports/` (e.g., cleaned data, reports, charts)

## How to use the exercises
1. **Read each exercise notebook** and follow the instructions step by step.
2. **Write your code** in the notebook cells or in the `src/python_rap_demo/` modules as instructed.
4. **Check your solutions** by comparing your results with the solutions notebooks in `exercises/solutions/`.
5. **View outputs** in the `data/outputs/` and `reports/` folders (e.g., cleaned data, markdown reports, charts).

## Adding code to the src folder
- Place reusable functions and classes in the appropriate module in `src/python_rap_demo/` (e.g., `cleaning.py`, `io.py`, `report.py`).
- Use clear function names, type hints, and docstrings following [PEP8](https://peps.python.org/pep-0008/) standards.

## How to check if your solutions have worked
View the solutions for each exercise in the `exercises/solutions/` folder. The solutions will walk through the expected outputs and how to run them. This includes:
- **Comparing outputs**: Check your results against the solutions notebooks and output files.
- **Checking outputs**: Review generated files in `data/outputs/` or `reports/` for expected results.
- **Running tests**: For unit test exercises you can run all tests with:
  ```cmd
  pytest tests
  ```
- **Using pre-commit hooks and CI**: Ensure your code passes formatting, linting, and CI checks.

## Troubleshooting
If you encounter issues:
- Ensure your virtual environment is activated. The terminal prompt should show `(.venv)` at the start.
- Check your notebook is using the correct Python kernel associated with the virtual environment. You can change the kernel in the Jupyter notebook interface.
- Clean outputs and restart the kernel if you face issues.
- Check that all dependencies are installed by running `pip install -r requirements.txt`.
- Verify you are in the project root directory when running commands. The terminal should show the path ending with `python_rap_demo`.
