# Exercise: Using modules and functions

In this exercise, you'll practice refactoring a single, monolithic script into a modular pipeline using functions and separate files. This is a key RAP best practice for reproducibility and maintainability.

**Learning objectives:**

The code below is for example purposes and allows you to run the code in Step 1:

```python
# Monolithic script (starter code)
import pandas as pd

# Load data
df = pd.read_csv("../data/input/health_data.csv")

# Clean data (drop rows with missing diagnosis)
cleaned = df.dropna(subset=["diagnosis"])

print("\nFirst 5 rows of cleaned data:")
print(cleaned.head())
```

## Step 1: Review the Monolithic Script

Below is a single script that loads health data, manipulates it, and performs imputation for missing values. Your task is to refactor this into modular code.

**Tasks:**
- Convert the below into a single function

**Hints:**

```python
# Impute missing height and weight with column means
cleaned["height_cm"] = cleaned["height_cm"].fillna(cleaned["height_cm"].mean())
cleaned["weight_kg"] = cleaned["weight_kg"].fillna(cleaned["weight_kg"].mean())
```
